```python
import requests
import json
import os
from dotenv import load_dotenv
# ─── Load environment variables from .env file ────────────────────────────────
load_dotenv(override=True)  # override=True allows .env values to overwrite existing env vars

# ─── Configuration ────────────────────────────────────────────────────────────
API_KEY = os.getenv("LLMGW_API_KEY")
ENDPOINT =os.getenv("LLMGW_BASE_URL")
MODEL = os.getenv("LLMGW_MODEL")

print (f"API KEY {API_KEY}")
print (f"ENDPOINT {ENDPOINT}")
print (f"MODEL {MODEL}")

# ─── Headers (same as Postman Headers tab) ────────────────────────────────────
headers = {
    "x-api-key": API_KEY,
    "anthropic-version": "2023-06-01",
    "content-type": "application/json"
}
# ─── Request Body (same as Postman Body → raw → JSON) ─────────────────────────
payload = {
    "model": MODEL,
    "max_tokens": 256,
    "temperature": 0.5,
    "system": "You are a helpful assistant.",
    "messages": [
        {
            "role": "user",
            "content": "What is the capital of France?"
        }
    ]
}

# ─── Send POST Request ─────────────────────────────────────────────────────────
response = requests.post(ENDPOINT, headers=headers, json=payload, verify=False)
# ─── Display Response ──────────────────────────────────────────────────────────
print(f"Status Code : {response.status_code}")
print(f"Response    :\n{json.dumps(response.json(), indent=2)}")
```
